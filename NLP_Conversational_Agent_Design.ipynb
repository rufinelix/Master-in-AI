# VIA, VectorAhorro conversational agent

## Introduction

VectorAhorro is a medium-sized bank-insurer with 1.4 million customers and a digital channel that concentrates the majority of interactions. Its call center manages 38,000 monthly contacts, with 60% repetitive queries, an average wait time of 7 minutes, and a CSAT of 6.1/10, which motivates the deployment of the conversational agent VIA on app, web, and WhatsApp Business. The report is organized into five blocks: agent analysis, conversational design, NLP/ML technology, implementation and ROI, and ethical-legal impact, following the tasks of the extraordinary route.

---

## Part 1. VIA agent analysis

### 1.1 Type of agent and PEAS

VIA is a goal-based agent with learning capabilities, as it not only reacts to messages, but also pursues explicit business goals (reduce repetitive load, improve times, comply with regulation, reduce friction). It perceives, decides, and acts guided by performance criteria, beyond simple rules. Compared to a static FAQ, VIA incorporates semantic interpretation, context, escalation policies, and tracking metrics, thus fitting into the category of intelligent agent.

VIA's PEAS is summarized as follows:

- **Performance**: reduction of repetitive contacts in the call center, improvement of response time, increase in CSAT and, inseparably, compliance with GDPR, MiFID II, IDD, and frameworks such as the EU AI Act.
- **Environment**: app, web, and WhatsApp Business, retail customers, banking and insurance systems, CRM, and normative documentary bases in a regulated and dynamic environment.
- **Actuators**: textual responses, guided options, links, ticket creation, and derivations to the human channel.
- **Sensors**: natural language messages, channel metadata, session context, and signals from internal systems (status of procedures, contracted products), together with emotion indicators when sentiment analysis techniques are incorporated.

The environment is partially observable, non-deterministic, and dynamic, because the agent only sees part of the user's state, the language is ambiguous, and the regulatory or product conditions change. The interaction is multi-agent (people + VIA), with relevant uncertainty about real intention and emotional tone.

### 1.2 Scope, exclusions and decision table

- **VIA resolves**:  
  - General status of procedures (for example, tracking of already registered requests and claims).  
  - Necessary documentation in general terms.  
  - General product information (without individualized recommendations).  
  - App and web navigation (locate functions, explain routes).

- **Escalates to human**:  
  - Financial or insurance advice subject to MiFID II/IDD.  
  - Sensitive claims or high-impact complaints.  
  - Orders or operations with economic effect.  
  - Cases with low confidence of the intent classifier.  
  - Interactions with relevant negative emotion (anger, threat of claim).

- **Out of domain**:  
  - Topics unrelated to VectorAhorro's offerings or its financial-insurance scope.  
  - Insecure requests (for example, sharing PINs or full credentials) that must be safely rejected.

| Query type                             | VIA responds | Escalates | Out of domain |
|---------------------------------------|-------------|--------|-------------------|
| Status of procedures, documentation, general product info, navigation help | Yes, general informative level. | If it leads to a claim or advice. | — |
| Advice, sensitive claims, operations with economic impact | No | Yes, mandatory escalation. | — |
| Insecure or out-of-scope requests | No | — | Safe rejection and redirection. |

This scheme applies the S4 criteria of the RAA (repetitive, frequent, high error tolerance) to a regulated domain and avoids the over-automation that session 14 points out as a common mistake.

## Part 2. Conversational design

### 2.1 Bot components and conversational persona

VIA's architecture is organized into six components:

1. **Interface**: presence in app, web and WhatsApp Business, with transparency messages, help elements and session management.
2. **NLP**: text normalization, language detection, intent detection (important for the agent's conversational strategy) and entity extraction using tools like spaCy or Hugging Face models.
3. **Dialog manager**: minimal context maintenance, action selection, fallback management and application of escalation rules (advice, low confidence, negative emotion, economic operations and out of domain).
4. **Knowledge base**: validated corpus of products, procedures and documentation, with version control and review by business and compliance.
5. **Integrations/API**: controlled access to internal systems (status queries, ticket opening, etc.), under the principle of least privilege.
6. **Analytics**: KPI dashboards (FCR, CSAT, recontact, net containment, correct escalation, model metrics), following the distinction between health metrics and vanity metrics.

The agent's **persona** is defined as professional and clear, prudent, explicitly non-advisory and transparent about its automated nature to the point of explicitly reminding that it is an intelligent agent, complying with ethical issues and responsible communication. It is a proactive, empathetic and fast-resolving assistant (oriented towards the goal of streamlining procedures), with a professional yet accessible tone, free of complex banking jargon. It stands out for its high adaptability, as it learns from each interaction to anticipate needs and customize solutions according to the client's profile. It maintains calm and technical precision in the face of massive query peaks, transmitting absolute security in critical operations. Its personality is that of an expert and patient company administrative agent, knows the business and offers this knowledge but never advises, simplifying banking and constantly evolving to offer increasingly efficient responses.

### 2.2 Intents, slots and basic flow. WhatsApp Business case-

As **intents**, they can be: `procedure_status_query`, `documentation_information`, `general_product_information`, `navigation_help`, `general_complaint`, `economic_operation_query`, `financial_advice`, `out_of_domain`.

The minimum **slots** are limited to what is essential: procedure type, product type, channel, incident type, requested document, urgency and an advice indicator (linguistic patterns like “what do you recommend”). This design respects the GDPR minimization principle and reduces unnecessary exposure.

**Synthetic flow**:

1. VIA introduces itself as an automated agent and declares its scope.
2. The client makes their query.
3. VIA classifies the intent and extracts basic entities.
4. If the case fits into “VIA resolves” and the confidence is sufficient, it responds with templates based on the knowledge base.
5. If information is missing, it requests minimum data; if the intent is critical or trust drops due to the wait, it deploys an escalation.
6. It records interaction metrics for later analysis.

In **WhatsApp Business**, furthermore, requesting identifying data without consent is avoided and it redirects to authenticated channels for any sensitive operation. It also avoids the risk of the client believing they are interacting with a human agent, which can more easily happen on this platform due to its everyday and personal use; therefore, it explicitly states that it is an intelligent agent, including detecting when the client treats it as if they believed it was a human to anticipate possible frustration.

### 2.3 Suggested platform

We suggest using **Rasa** as the platform, because it allows on-premise deployment or on infrastructures under the control of the entity, reinforcing GDPR and confidentiality; it integrates NLU, dialogue management, and escalation policies configurable with a high degree of detail; and it couples well with modern NLP models, maintaining traceability and audit capability. Although it implies a higher cost because it would require experts in programming intelligent agents, the volume of queries justifies this extra cost which will be reinvested directly and indirectly in the company in an economic and also reputational way.

## Part 3. NLP/ML Technology 

### 3.1 Intents, entities, and models

In VIA, the classifier must cover at least the intents defined in part 2 of this document and operate on labeled examples from the financial domain. For **intents**, supervised models can be used (for example, neural networks on embeddings or fine-tuned LLMs like BERT), trained with typical phrases of banking and insurance clients that reflect, for example, the previously proposed intents. For their part, the relevant **entities** include product type, procedure type, incident type, channel, requested document, operation, and advice marker, which can be extracted with spaCy NER or fine-tuned transformers for the financial context.

Out-of-domain messages are channeled to a specific class. In case the confidence is low among several classes, the bot must ask for clarification or escalate to avoid risky decisions.

### 3.2 Metrics, critical error, and mitigation

We would evaluate the classifier with accuracy (how many predictions it got right), precision (of all detected positives how many were real), recall (of all real positives how many I have detected), and F1 (the harmonic mean of the previous two used to be optimized to achieve a joint optimization of precision and recall). But in a regulated environment, the focus is not on the global average, but on the behavior per class, especially in `asesoramiento_financiero`. For example:
- The **most costly error**, in that class, is the false negative of the advice detector: a case that should be escalated and remains in the bot, exposing VectorAhorro to MiFID/IDD non-compliance and claims.
- The **limitations** facing this scenario include data scarcity in critical classes, training biases, linguistic ambiguity, regulatory changes, out-of-domain queries, and possible hallucinations if generative models are used.

Main solutions in this situation:

- Conservative confidence thresholds for sensitive intents (advice, economic operations).  
- Periodic human review of logs and edge cases.  
- Auditable records of decisions, consistent with GDPR and the EU AI Act.  
- Controlled knowledge base, avoiding generated responses without documentary backing.  
- Recurrent testing after product and regulatory changes, to anticipate drift.

It is also necessary to highlight a special metric which is trust; it cannot be calculated directly but it must be maintained high with clients to avoid a bad reputation that could cause the implementation of the agent to result in losses due to the cost of designing it and the lack of return on investment from the disuse that its ineffectiveness would cause.

## Part 4. Implementation and ROI 

### 4.1 ROI

We start from reasonable values for economic costs:
* Average cost per human query in banking: 9.50€ (includes salaries with banking collective agreements, bonuses, corporate software tools, and structure).
* Estimated cost of the enterprise intelligent agent: 4,500€/month (for this volume of queries in a secure banking environment with regulatory compliance).

On the other hand, the queries absorbed by the agent can be:
* Total Queries: 38,000 per month
* Automatable Queries (60%): 22,800 queries
* Estimated success/absorption rate of the agent: 85% of that 60% (that is, the agent successfully resolves 19,380 queries and escalates the rest to humans due to complexity).

| Concept | Calculation | Result |
| --- | --- | --- |
| Previous monthly expense (100% Human) | $38,000 \text{ queries} \times 9.50\text{€}$ | **361,000€** |
| Volume fully absorbed by AI | $22,800 \times 85\%$ | 19,380 queries |
| Gross Monthly Savings | $19,380 \text{ queries} \times 9.50\text{€}$ | **184,110€** |
| Monthly Investment in the agent (Total cost)| Licenses, infrastructure, and security | **4,500€** |

$$\text{Net Monthly Savings} = 184,110\text{€} - 4,500\text{€} = 179,610\text{€}$$

$$ROI = \frac{184,110\text{€} - 4,500\text{€}}{4,500\text{€}} \times 100 = \mathbf{3,991\%}$$

But there are also benefits that cannot be measured with ROI, at least not with a manual one like this (there are specialized calculators on the internet that do so with estimates), such as the improvement in waiting time that we expect to be reduced from 7 minutes to seconds and the customer satisfaction or CSAT that we expect to increase from 6.1 to a minimum of 8.

In summary, in annual terms there would be an estimated annual investment of 54,000 euros, but with a net annual savings of more than two million euros, 2,155,320 euros specifically, being also noteworthy the release of 135,660 minutes of waiting time every month that can be used in other higher value-added operations of the company.

These data are very good, but they must be taken with caution, as there are really sensitive and uncertain estimates such as the absorption rate that cannot be known with certainty until it is implemented and that the only way to ensure that it is high is a high investment in the design which, in any case, would not control all the aspects that could cause the trust in the agent to be low and not be used. We start from the idea that it will not be a disaster but for this it is advisable to look at the design of the agent with the objective of it being efficient over profitable, as an efficiency failure can be worse than the savings with a more cautious investment.

### 4.2 KPIs, GDPR, and financial compliance

We distinguish health metrics (expecting FCR between 65 to 75%, CSAT 78-85%, low recontact because we expect to resolve queries on the first try, net containment (retention in the customer's conversation but providing value to them) of 85%, correct escalation rate) from vanity metrics (volume of conversations, messages, gross containment (customer retention but without solving anything in some cases) without context). VIA must prioritize monitoring the former to avoid validating a busy bot that appears functional but is actually useless.

From GDPR and frameworks like MiFID II (on investment products), PCI DSS (on privacy), and the emerging EU AI Act, the system must:

- Base data processing on contract execution and legitimate interest, completed with consent when data is used for training.
- Minimize collected data, limit the use of unauthenticated channels, and anonymize training records. Let's remember privacy-by-design, this must be taken into account from the first line of code that programs the agent.
- Ensure traceability, right to human intervention, and clear claim channels.

The key tension between over-escalating, increasing costs but reducing risk, and under-escalating, saving in the short term but potentially generating sanctions and loss of trust, comes from not being able to resolve how the implementation of agents can affect the company's growth, because as much as it serves to resolve queries and streamline, it is necessary to have a total benefit, not just per client, that allows justifying the investment, this is where the current situation among financial institutions comes in, which maintain a lively and aggressive fight to get the largest number of clients, since as everyone needs a bank, this comes down to capturing dissatisfied clients and retaining them, something that can be caused by the reputation that may derive from a bad service unresolved by intelligent agents, but which could also be solved by these same agents if they were efficient enough. Of course, above all, the reasonable strategy in banking as a starting point is to prioritize client protection over aggressive ROI maximization.

## Part 5. Impact, ethics and recommendation

### 5.1 Transparency, risks and safeguards

VIA's ethical design demands transparency about the artificial nature of the agent, especially on WhatsApp Business, where the user might assume they are talking to a person due to the everyday and personal-private use nature of this platform. The type of help it offers, the limitations, and the possibility of referral to human channels must be made explicit.

As dictated by the GDPR, the user must be able to give their consent in an informed, clear, and active manner to use their data, avoiding pre-ticked boxes for such consent; the right to be forgotten is also important, that is, that the company can guarantee that the data of clients who request it are permanently deleted, and also that the client can access what data the company has about them; the agent should ask only for the data it needs depending on at what point in the conversation it is, for a greeting it should not ask, for example, for an income tax return, the agent must minimize the collected data to what is strictly necessary and respect privacy; the user must be informed of the agent's competencies to avoid situations of hallucinations in contexts where it was not trained because it crosses a red line as is the case of financial advice.

Main ethical-social risks:

- Incorrect or hallucinated answers in a financial context, especially dangerous if not cut off to avoid a financial advisory attitude. 
- Exclusion or added difficulty for less digital users.  
- Excessive trust in an unauthorized system to advise.  
- Claims before the Bank of Spain, DGSFP, or AEPD due to design or governance failures.

Recommended safeguards:

- Clear messages of transparency and response limits.  
- Conservative escalation policy for sensitive cases.  
- Periodic audit of conversations and content review.  
- Accessible alternative channels (phone, offices, human chat) and accessibility measures such as education with video tutorials on how to interact with the agent for those with less digital education.

### 5.2 Synthesis and recommendation to the Risk Committee

In summary, VIA appears as a goal-oriented conversational agent, with learning elements and a PEAS where efficiency, compliance, and client protection form a single performance block. The conversational design delimits a clear scope, defines escalation triggers, and relies on an architecture aligned with its objectives. The NLP/ML framework prioritizes the reliable detection of advice and critical queries, accepting over-escalation as the price of prudence. In business terms, the deflection hypothesis is reasonable, but viability depends on not overestimating containment nor underestimating governance costs. Ethically and legally, VIA is only acceptable if it respects GDPR, MiFID II, IDD, and the EU AI Act guidelines; above all, maintaining privacy-by-design is a non-negotiable minimum from the beginning.

The recommendation to the Risk Committee is to **deploy VIA with conditions**:

- Scope limited to informational and low-risk operational queries.  
- Mandatory escalation in the face of advice, negative emotion, economic operations, low confidence, and out-of-domain.  
- Robust governance: human review, auditable logs, health KPIs, and joint supervision of business, technology, and compliance.

Under these conditions, VIA can improve customer service without compromising their protection; without them, the project would drift into apparent automation with high regulatory and reputational risk that would end in losses due to the cost of designing it and the non-return in benefits that did not arrive due to the disuse of the agent caused by its inefficiency.
