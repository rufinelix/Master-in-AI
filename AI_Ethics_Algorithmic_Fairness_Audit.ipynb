# Ethical audit of automated personnel selection systems with biometrics: analysis of the HireVue case

## Introduction and delimitation of the case

The selected case is that of automated personnel selection systems that, in addition to classifying resumes and prioritizing applications, may incorporate biometric analysis to infer personality traits, suitability, or soft skills during video interviews. HireVue is taken as an illustrative reference, a company known for having marketed video interview and automated evaluation solutions in hiring processes. Although it removed the facial analysis component, the case remains useful for analyzing the ethical dilemmas in the field. In any case, we are not limiting ourselves to a specific company, but rather examining the possible and hypothetical AI systems used in human resources to filter CVs, score interviews, rank applicants, or support hiring, promotion, and exclusion decisions. In the current European framework, these uses are particularly sensitive because the AI Act considers systems used in employment access, labor management, and self-employment to be high-risk, while also prohibiting certain biometric uses such as emotion recognition in the workplace and in educational contexts.

## Task 1. Description of the case and the AI system

### Name, organization and application context

The theoretical system analyzed is an AI-assisted automated selection platform for corporate recruitment. HireVue, for example, offered video interviews and automated evaluations to companies wishing to reduce costs and scale globally. The application sector is human resources in hiring processes with a high volume of applicants, especially in large companies, consulting firms, financial institutions, retail chains, and organizations seeking to automate tasks such as receiving applications, resume analysis, and pre-selection based on patterns detected by algorithms.

### System purpose

The main purpose of the system is to optimize decision-making in the early stages of recruitment, automating or relying on actions such as filtering CVs, classifying profiles according to predetermined criteria, assigning scores to recorded interviews, and generating applicant rankings for later review by human personnel of the most relevant cases. When the system incorporates biometrics, the automation goes beyond documentary analysis: it can process voice, tone, speech rate, facial micro-expressions, posture, gaze, or other behavioral indicators to infer attributes such as confidence, sociability, emotional stability, motivation, or cultural fit, although this class of inferences is scientifically controversial and ethically problematic.

### Users and affected groups

The direct users of the system are usually recruiters, human resources departments, and hiring managers. However, the main affected, those harmed, are those who apply for the jobs since the system can condition their access to employment, reduce their visibility in the process, and decisively affect their job opportunities. As a special case of bias, other groups such as people with disabilities, ethnic or linguistic minorities and, in general, any social group whose representation in historical data is limited or distorted, are indirectly affected, as long as this imbalance is not resolved when training the models.

### Type of data used

These systems can use a combination of personal and professional data declared in the resume: academic and professional history, declared skills, previous experience, and others obtained in later phases of the selection process, such as responses to questionnaires, psychometric test results, and process metadata. When video analysis is involved, biometric or biometrically inferred data are added, such as facial images, voice patterns, linguistic features, and other behavioral indicators extracted from the conducted interviews. These data may seem harmless, but they can encourage the system to attempt to deduce intimate or protected traits from seemingly neutral data. It should be mentioned that even anonymization may not prevent re-identification or discriminatory impact.

### Expected benefits

The expected benefits include reduced hiring time, interview standardization, management of large volumes of applications, a more consistent evaluation (and even fair by avoiding human bias at the cost of the artificial bias of the systems) than purely manual evaluation, and an improvement in operational traceability, language skill detection, and the efficiency of human resources teams. From a business perspective, automation can reduce costs and accelerate the selection funnel. Furthermore, a well-designed tool could help decrease some individual recruiter biases if limited to relevant, auditable, and contextually justified criteria.

### Initial risks

The initial risks of the case are high. As mentioned, AI systems can amplify historical discrimination when they learn from biased data or when design decisions make certain groups invisible. In personnel selection, this problem is especially serious because the decision affects access to work, which is a central dimension of social inclusion and equal opportunities. Without forgetting the risks of privacy, opacity, defenselessness of the candidate, use of biometric inferences without robust basis, excessive surveillance, and weakening of real human supervision. All this negatively affects the company's reputation, potentially leading to a lack of candidates if they do not see the company as a safe and trustworthy environment, which would nip a company's growth in the bud, something critical from a business point of view if we take into account that those seeking to absorb more employees or select them better are because they are growing.

## Task 2. Ethical analysis of the case

### Deontological perspective

From deontology, the ethical examination starts from duties, rights, and limits that should not be violated even if the tool increases efficiency. The first of these limits is respect for human dignity: a candidate should not be treated as a mere set of computable signals or evaluated through invasive inferences about their personality from questionable biometrics. The right to equality and non-discrimination also comes into play: if the system systematically disadvantages women, racial minorities, neurodivergent individuals, people with disabilities, or speakers with certain accents, a basic duty of procedural justice is violated, even if the organization had no explicit intention to discriminate. Another essential deontological duty is privacy protection, as AI intensifies the risks of sensitive data inference, requiring the application of minimization, transparency, and proactive responsibility principles in line with the GDPR and the Spanish legal framework. Consequently, capturing or processing video, voice, and behavioral traits would only be defensible if there were a strict need, a solid legal basis, and proportionality guarantees that, in many recruitment scenarios, are difficult to justify. For this very reason, HireVue, among other reasons, had to stop marketing its system integrating facial analysis. Finally, deontology also demands sufficient transparency and the possibility of challenge. Placing ourselves in context, when an algorithmic decision or recommendation seriously conditions access to employment, the affected person must understand that this automated intervention exists, what general logic guides it, and how they can request meaningful human review; conversely, if the system works as a black box, the organization breaches duties of information, justification, and respect.

### Utilitarian perspective

From utilitarianism, it would be necessary to assess whether the system generates more benefits than harms for the group of affected people; although it should be said that certain harms could invalidate the greatest of benefits, of course. At first glance, a platform like HireVue can offer appreciable utilities: time savings, cost reduction, operational standardization, and the ability to manage massive selection processes. However, the utilitarian calculation cannot be limited to the user company, but must also include the potential harms suffered by unjustly excluded candidates, by structurally biased groups, and by society itself if automation consolidates labor inequalities under technical appearance. When the system uses historical hiring data, there is a serious risk of reproducing past preferences and converting them into automated criteria in future cases. A documented case in this regard is the Amazon case, which used a recruitment system that penalized women, illustrating how AI can amplify injustices with an appearance of objectivity. If this happens, the efficiency benefit for the company is morally weakened by the accumulated damage to equality and social trust. In addition, on the utilitarian plane, the empirical reliability of inferences matters; that is, if facial or linguistic analysis promises to measure traits such as emotional intelligence or cultural fit without sufficient scientific validation, the system can produce a large volume of errors with severe consequences on specific career paths. In such a case, the welfare balance would be negative, because a slight gain in organizational productivity does not compensate for unfair labor exclusions, reputational damage, and loss of institutional legitimacy.

### Virtue ethics and professional responsibility

Virtue ethics shifts the focus of the ethical analysis towards the moral character and prudence of those who design, hire, and supervise these systems; that is, it analyzes the ethics of the system through the individual and organizational ethics of those responsible for it. In this case, relevant virtues would be justice, prudence, intellectual honesty, responsibility, empathy, and epistemic humility regarding what technology can truly know about a person. Specifically, designers and providers should act with methodological honesty, which implies not exaggerating the predictive capabilities of the system, not presenting weak correlations as robust psychological traits, and not hiding uncertainties that directly affect fundamental rights under technical language. For their part, managers and human resources managers should adopt an attitude of prudence and professional responsibility away from simple immediate economic benefit, since it is not enough to buy a tool because it promises efficiency; they must require audits, validate results by subgroups, verify the proportionality of data processing, and ensure that the final decision is not handed over to opaque automatisms. From this perspective, a virtuous organization does not simply formally comply with the rule, it also cultivates critical review and is open to complaints, participation of affected individuals, and active correction of unjust impacts.

### Biases, discrimination and negative impacts

The first material risk is algorithmic bias, which stems both from training data reflecting historical discrimination and from design decisions making minorities invisible. In personnel selection, this can translate into indirect penalization by gender, age, origin, disability, communicative style, university of origin, or employment gaps, even when sensitive variables are not explicitly introduced, since, as already mentioned, the system may attempt to infer them. The second risk is biometric and behavioral discrimination, that is, systems that analyze facial expressions, gaze, tone, or prosody can unfairly punish autistic people, those with facial paralysis, anxiety, stuttering, non-normative accents, or distinct cultural communication styles. This objection is not marginal, because it precisely affects groups that already usually encounter additional barriers to accessing employment. A third negative impact is the naturalization of opaque criteria of "cultural fit" or "potential", categories that could function as vehicles for subjective or class biases, but with a more neutral appearance because they are mediated by AI. The result of these risks translates into decisions with a high social impact supported by criteria difficult to audit and discuss.

### Privacy, transparency, explainability and human oversight

Privacy is a critical axis of the case. In a selection process, asking thousands of applicants to record their face and voice to extract behavioral traits represents an intrusion far superior to the classic evaluation of a CV. Furthermore, as already mentioned, it is necessary to prevent the inference of sensitive data from other apparently neutral ones.

Transparency cannot be understood superficially either. It is not enough to report the use of AI; it is necessary to explain comprehensively what it is used for, what variables count, what relative weight they have, what limits it works with, and how possible errors are reviewed. Without that level of clarity, the explanation becomes an obfuscation of truth, if not a lie, by appearing to be accountable without properly doing so.

Regarding human supervision, the European regulation on high-risk AI, as is the case of personal data and contexts having a direct impact on people, requires appropriate measures of human control, traceability, documentation, data quality, and risk mitigation. In addition, the AI Act considers AI tools used in employment and recruitment, such as CV screening software, to be high risk, and prohibits emotion recognition in workplaces and educational environments. In short, in personnel selection, a symbolic review is not enough; meaningful human supervision is required, with real capacity to correct, halt, or override the algorithmic recommendation.

### Overall ethical assessment

With all the ethical frameworks analyzed, the system cannot be considered ethically acceptable in its most intrusive form, especially when it includes biometrics to infer personality or emotions. In the current European context, emotion recognition in employment is prohibited, and the use of AI in recruitment falls into the high-risk category, subject to enhanced requirements. Additionally, the risk for all involved, both candidates and company, is high, and the cost too. Therefore, this type of system could only be acceptable under very strict conditions and, regarding emotional or biometric personality inferences, should be completely rejected. A much more limited use of non-biometric automation could be considered, limited to organizational support, with relevant data, independent auditing, effective transparency, the right to review, and the exclusion of discriminatory variables.

## Task 4. Proposal for responsible governance

### Guiding ethical principles

The governance of the system should be based on a set of principles consistent with international legal and moral frameworks: human dignity, justice and non-discrimination, privacy, transparency, explainability, traceability, security, and meaningful human supervision. These principles should not remain rhetorical statements; they must translate into design decisions (privacy-by-design), hiring, validation, and system withdrawal. Likewise, it is advisable to adopt the criterion that the burden of justification must be particularly high and that any intrusive functionality must be assumed beforehand as problematic until its strict necessity and reliability are proven.

### Transparency and explainability measures

The responsible organization should inform from the beginning that algorithmic support exists in the selection process, indicating purposes, categories of processed data, general evaluation criteria, and candidate rights. The CDT report on the AI Act recalls that, for high-risk employment systems, there are information obligations to workers and affected persons, as well as explanation and complaint mechanisms in certain cases. On an operational level, accessible documentation should be generated with at least five elements: variables used, variables excluded, model purpose, global and subgroup metrics, and known system limits. In addition, every automated recommendation should be accompanied by an understandable justification for the human resources system personnel, avoiding its black box nature as the exclusive basis for decision.

### Meaningful human oversight

Meaningful human oversight requires much more than a final signature; it implies that a process must be designed in which the reviewer can understand the system's output, detect inconsistencies, access additional contextual evidence, and modify or annul the recommendation without organizational friction. For this, it is advisable to establish concrete rules: prohibition of fully automated rejection, mandatory review of borderline cases, priority review of complaints, and a registry of when and why a human reviewer confirms or corrects the algorithmic output. This requirement connects with the idea, present in both course materials and UNESCO, that AI must not displace ultimate human responsibility.

### Complaint and review procedure

The company or organization should enable a specific, visible, and free complaint procedure, in which any candidate should be able to request information on the use of AI in their evaluation, request human review of the decision, and question the relevance of certain data or inferences. This procedure should include short deadlines, review by personnel other than the initial evaluator, and documentary traceability. And when the system is high risk, the right to a clear and meaningful explanation takes on special relevance to prevent the challenge from being purely formal.

### Fair and safe evaluation indicators

Responsible governance needs measurable indicators such as the false negative rate by subgroup, selection differences between gender and origin, consistency between human and algorithmic evaluations, number of accepted complaints, average review time, and percentage of decisions corrected after human supervision. Privacy and compliance indicators should also be monitored, such as data volume collected, retention time, security incidents, unauthorized access, and number of sensitive variables.

Monitoring should not be ad hoc, but continuous and with internal, external and, where possible, participatory audits, to mitigate bias and support inclusive governance.

### Recommendations applicable to the organization

Based on the analysis conducted, the main recommendations would be the following:

- Eliminate any emotion recognition functionality or personality trait inference (both from biometrics and in other cases, having to impose firewalls on this type of inferences), due to its low ethical legitimacy and the prohibitions and restrictions existing in the European framework.
- Limit AI to administrative support tasks or preliminary prioritization with variables strictly related to the job and empirically validated.
- Conduct an impact assessment on fundamental rights and data protection prior to deployment, even when no explicit legal obligation exists for all private cases.
- Require internal, external, and participatory, independent audits on bias, performance, and explainability, including validating the system by subgroups and publishing understandable summaries of results.
- Implement meaningful human supervision, with specific training for human resources and real capacity for review and rectification.
- Offer alternative non-biometric evaluation pathways to guarantee accessibility and reasonable accommodations in cases of disability, neurodivergence, or linguistic barriers.
- Include contractual clauses with the provider guaranteeing access to documentation, metrics, incidents, model changes, and shared responsibilities.

## Reasoned final conclusion

The analysis of the case shows that automated personnel selection systems constitute one of the most delicate scenarios of artificial intelligence ethics, because they combine a direct impact on fundamental rights and a strong temptation to delegate complex decisions to opaque metrics. The problem is aggravated when the evaluation incorporates biometrics to infer personality, emotions, or suitability, since there the low epistemological reliability of the inference, the intrusion into private life, and a high risk of indirect discrimination intersect. From deontology, the system fails when it objectifies the candidate, compromises their privacy, or hinders the right to a fair review. From utilitarianism, the supposed efficiency benefits do not compensate for the accumulated harms if the system reproduces unfair exclusions or relies on weak scientific foundations. From virtue ethics, a responsible organization should act with prudence, methodological honesty, and active commitment to equity and accountability. In conclusion, the implementation of AI systems in personnel selection would only be defensible if their function is strictly limited to empirically proven preliminary and organizational tasks, is audited, explainable, proportional, and subjected to meaningful human control; instead, biometric functionalities, or not, aimed at inferring emotions or personality must always be rejected for ethical and regulatory reasons. In terms of governance, the most reasonable decision is not to accept the technology as offered, but to profoundly redesign it under a rights-based approach, data minimization, continuous auditing, and reinforced organizational responsibility, in short, privacy-by-design.
